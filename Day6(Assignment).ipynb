### Imports
We begin by importing all the necessary libraries for our preprocessing tasks.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder

print("Libraries imported successfully!")
```

### Create Sample Dataset


```python
raw_data = {
    'Age': [22, 25, 47, 35, np.nan, 28, 55, 60, 24, 31],
    'Salary': [45000, 52000, 110000, 75000, 62000, 58000, 125000, 250000, np.nan, 300000],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male', np.nan, 'Male', 'Female', 'Female'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'New York', 'Chicago', 'Los Angeles', 'New York', 'Chicago', 'Los Angeles', np.nan],
    'Date': ['2023-01-15', '2023-01-16', '2023-02-20', '2023-03-12', '2023-04-05', '2023-05-18', '2023-06-22', '2023-07-01', '2023-08-11', '2023-09-05'],
    'Purchase_Amount': [120.50, 85.00, 450.25, 300.00, 50.00, 150.75, 520.00, 1200.00, 95.00, 210.00],
    'Experience': [1, 2, 15, 8, 4, 5, 20, 25, 2, 6],
    'Rating': [3.5, 4.0, 4.5, 4.2, 3.8, 3.9, 4.8, 5.0, 3.7, 4.1]
}

df_original = pd.DataFrame(raw_data)
df = df_original.copy()

print("Original Dataset Summary:")
display(df.info())
display(df.head(10))
```

### Problem 1: Handling Missing Values
We will fill numerical columns with their mean and categorical columns with their mode.

```python
for col in ['Age', 'Salary']:
    mean_value = df[col].mean()
    df[col] = df[col].fillna(mean_value)

for col in ['Gender', 'City']:
    mode_value = df[col].mode()[0]
    df[col] = df[col].fillna(mode_value)

print("Missing values per column after imputation:")
print(df.isnull().sum())
display(df.head(10))
```

### Problem 2: Feature Scaling
We apply `StandardScaler` to Salary and `MinMaxScaler` to Age, then display the comparisons.

```python
scaler_standard = StandardScaler()
df['Salary_Scaled'] = scaler_standard.fit_transform(df[['Salary']])

scaler_minmax = MinMaxScaler()
df['Age_Scaled'] = scaler_minmax.fit_transform(df[['Age']])

display(df[['Age', 'Age_Scaled', 'Salary', 'Salary_Scaled']].head(10))
```

### Problem 3: Encoding Categorical Data
We transform categorical labels to numerical representations using `LabelEncoder` for Gender and `pd.get_dummies()` for City.

```python
encoder = LabelEncoder()
df['Gender_Encoded'] = encoder.fit_transform(df['Gender'])

city_dummies = pd.get_dummies(df['City'], prefix='City', dtype=int)
df = pd.concat([df, city_dummies], axis=1)

display(df.head(10))
```

### Problem 4: Feature Creation
We create a new feature named `Total_Spending` and explain its usage.

```python
df['Total_Spending'] = df['Salary'] + df['Purchase_Amount']

display(df[['Salary', 'Purchase_Amount', 'Total_Spending']].head(10))
```

### Problem 5: Date Feature Extraction
We parse the Date column and extract useful subcomponents.

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.weekday

display(df[['Date', 'Year', 'Month', 'Day', 'Weekday']].head(10))
```

### Problem 6: Binning
We group the age values into distinct groups using `pd.cut()`.

```python
bins = [0, 25, 45, float('inf')]
labels = ['Young', 'Adult', 'Senior']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

display(df[['Age', 'Age_Group']].head(10))
```

### Problem 7: Outlier Detection
We identify salary outliers using the Interquartile Range (IQR) method.

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]

print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
print("Outlier Rows:")
display(outliers)
```

### Problem 8: Feature Interaction
We create a multiplicative feature combining `Experience` and `Salary`.

```python
df['Experience_Salary'] = df['Experience'] * df['Salary']

display(df[['Experience', 'Salary', 'Experience_Salary']].head(10))
```

### Problem 9: Feature Selection
We calculate numeric correlation coefficients, rank them, find the weakest feature, and drop it.

```python
numeric_cols = df.select_dtypes(include=[np.number])

correlations = numeric_cols.corr()['Purchase_Amount'].sort_values(ascending=False)
print("Correlations with Purchase_Amount in descending order:")
print(correlations)

correlations_no_target = correlations.drop('Purchase_Amount')
weakest_feature = correlations_no_target.abs().idxmin()
weakest_corr_value = correlations_no_target[weakest_feature]

print(f"\nFeature to drop (lowest absolute correlation): '{weakest_feature}' (Correlation: {weakest_corr_value:.4f})")

df_final = df.drop(columns=[weakest_feature])
print("\nFinal Dataset Sample after dropping the weakest feature:")
display(df_final.head(10))
```
