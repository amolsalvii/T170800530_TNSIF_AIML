```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="darkgrid")
```

## Task 1: Matplotlib Line Graph

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [200, 300, 250, 400]

plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker='o', color='b', linestyle='-', linewidth=2, markersize=8)

plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Monthly Sales Performance')
plt.grid(True)
plt.show()
```

## Task 2: Seaborn Functions Demonstration

### 1. sns.scatterplot() — Scatter plot

```python
df_scatter = pd.DataFrame({
    'Study Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Exam Score': [50, 55, 65, 70, 75, 85, 90, 95]
})

plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_scatter, x='Study Hours', y='Exam Score', color='blue', s=100)
plt.title('Study Hours vs. Exam Score')
plt.show()
```

### 2. sns.lineplot() — Line plot

```python
df_line = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021, 2022],
    'Company Growth (%)': [5, 12, 8, 15, 22]
})

plt.figure(figsize=(6, 4))
sns.lineplot(data=df_line, x='Year', y='Company Growth (%)', marker='s', color='green', linewidth=2.5)
plt.title('Company Growth Over Years')
plt.show()
```

### 3. sns.barplot() — Bar chart

```python
df_bar = pd.DataFrame({
    'Product': ['Apples', 'Bananas', 'Oranges', 'Berries'],
    'Quantity Sold': [120, 180, 90, 150]
})

plt.figure(figsize=(6, 4))
sns.barplot(data=df_bar, x='Product', y='Quantity Sold', palette='viridis', hue='Product', legend=False)
plt.title('Quantity of Products Sold')
plt.show()
```

### 4. sns.countplot() — Category counts

```python
df_count = pd.DataFrame({
    'Favorite Fruit': ['Apple', 'Banana', 'Apple', 'Orange', 'Banana', 'Banana', 'Apple', 'Apple']
})

plt.figure(figsize=(6, 4))
sns.countplot(data=df_count, x='Favorite Fruit', palette='Set2', hue='Favorite Fruit', legend=False)
plt.title('Count of Favorite Fruits')
plt.show()
```

### 5. sns.boxplot() — Data distribution

```python
df_box = pd.DataFrame({
    'Class': ['Class A']*5 + ['Class B']*5,
    'Test Score': [85, 90, 78, 92, 88, 70, 75, 80, 68, 82]
})

plt.figure(figsize=(6, 4))
sns.boxplot(data=df_box, x='Class', y='Test Score', palette='pastel')
plt.title('Test Score Distribution by Class')
plt.show()
```

### 6. sns.violinplot() — Distribution by category

```python
df_violin = pd.DataFrame({
    'Group': ['Group X']*10 + ['Group Y']*10,
    'Age': [23, 25, 22, 24, 28, 21, 25, 26, 24, 23, 35, 38, 32, 40, 31, 36, 34, 39, 35, 37]
})

plt.figure(figsize=(6, 4))
sns.violinplot(data=df_violin, x='Group', y='Age', palette='muted')
plt.title('Age Distribution by Group')
plt.show()
```

### 7. sns.heatmap() — Correlation heatmap

```python
data_corr = pd.DataFrame({
    'Math': [80, 90, 75, 60, 95],
    'Science': [85, 88, 80, 65, 90],
    'History': [70, 65, 80, 85, 60]
})

plt.figure(figsize=(6, 4))
sns.heatmap(data_corr.corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Subject Performance Correlation')
plt.show()
```

### 8. sns.pairplot() — Relationships between variables

```python
df_pair = pd.DataFrame({
    'Height (cm)': [150, 160, 170, 180, 190],
    'Weight (kg)': [50, 60, 68, 75, 85],
    'Gender': ['F', 'F', 'M', 'M', 'M']
})

sns.pairplot(df_pair, hue='Gender', palette='coolwarm')
plt.suptitle('Height and Weight Relationships', y=1.02)
plt.show()
```

### 9. sns.histplot() — Histogram

```python
scores = [55, 62, 67, 71, 73, 75, 78, 81, 85, 88, 90, 92, 95, 98]

plt.figure(figsize=(6, 4))
sns.histplot(scores, bins=5, kde=True, color='purple')
plt.xlabel('Exam Scores')
plt.title('Distribution of Exam Scores')
plt.show()
```

### 10. sns.regplot() — Regression plot

```python
df_reg = pd.DataFrame({
    'Advertising Spend ($)': [100, 200, 300, 400, 500, 600],
    'Product Revenue ($)': [1200, 1500, 1900, 2100, 2600, 3000]
})

plt.figure(figsize=(6, 4))
sns.regplot(data=df_reg, x='Advertising Spend ($)', y='Product Revenue ($)', color='red')
plt.title('Advertising Spend vs. Product Revenue')
plt.show()
```

### 11. sns.catplot() — Categorical plot

```python
df_cat = pd.DataFrame({
    'Day': ['Thur', 'Fri', 'Sat', 'Sun'] * 3,
    'Total Bill': [15.5, 18.2, 25.1, 28.4, 14.3, 19.8, 22.5, 30.1, 16.2, 17.5, 26.4, 29.5],
    'Time': ['Lunch', 'Lunch', 'Dinner', 'Dinner'] * 3
})

g = sns.catplot(data=df_cat, x='Day', y='Total Bill', hue='Time', kind='swarm', palette='Set1', legend=True)
g.fig.suptitle('Total Bill Amount by Day and Meal Time', y=1.02)
plt.show()
```
