### Load Dataset
Upload your dataset CSV file and load it into a pandas DataFrame.

```python
import pandas as pd
import numpy as np
from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_csv(filename)

print(df.head())
```

### Clean Column Names
Convert all column names to lowercase and replace spaces with underscores for easier coding.

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
print("Cleaned Columns:")
print(df.columns)
```

### 1. Display First 5 and Last 5 Rows

```python
print("--- First 5 Rows ---")
display(df.head(5))

print("\n--- Last 5 Rows ---")
display(df.tail(5))
```

### 2. Find Total Number of Rows and Columns

```python
rows, cols = df.shape
print(f"Total Rows: {rows}")
print(f"Total Columns: {cols}")
```

### 3. Display Data Types of Each Column

```python
print(df.dtypes)
```

### 4. Identify and Count Missing Values

```python
print(df.isnull().sum())
```

### 5. Fill Missing Values
We fill numerical columns with their mean, and categorical columns with their mode.

```python
for col in df.columns:
    if df[col].dtype in [np.float64, np.int64]:
        df[col] = df[col].fillna(df[col].mean())
    else:
        if not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after filling:")
print(df.isnull().sum())
```

### 6. Remove Duplicate Rows

```python
df = df.drop_duplicates()
print(f"Remaining rows after removing duplicates: {df.shape[0]}")
```

### 7. Convert Date Column to Datetime
Check if a date column exists, otherwise explain that it does not.

```python
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])
    print("Successfully converted 'date' to datetime.")
else:
    print("This dataset does not contain a 'date' column.")
```

### 8. Convert Floating-Point Columns to Integers
Convert numeric columns (like scores) to integers where safe and applicable.

```python
for col in ['math_score', 'reading_score', 'writing_score']:
    if col in df.columns:
        df[col] = df[col].astype(int)

print(df[['math_score', 'reading_score', 'writing_score']].dtypes)
```

### 9. Select Multiple Columns and Fill Missing Values Together
Demonstration of filling missing values in multiple selected columns using a dictionary mapping.

```python
cols_to_fill = [col for col in ['math_score', 'reading_score', 'writing_score'] if col in df.columns]
if cols_to_fill:
    fill_values = {col: df[col].mean() for col in cols_to_fill}
    df_temp = df.fillna(value=fill_values)
    print("Filled selected columns successfully.")
```

### 10. Increase Math, Reading, and Writing Scores by 10%
We create new columns for the updated scores.

```python
if 'math_score' in df.columns:
    df['math_score_plus_10'] = df['math_score'] * 1.10
if 'reading_score' in df.columns:
    df['reading_score_plus_10'] = df['reading_score'] * 1.10
if 'writing_score' in df.columns:
    df['writing_score_plus_10'] = df['writing_score'] * 1.10

display(df[[col for col in df.columns if 'plus_10' in col]].head())
```

### 11. Standardize Categorical Text
Standardize text values to lowercase and remove outer spaces.

```python
categorical_cols = ['gender', 'race/ethnicity', 'parental_level_of_education', 'lunch', 'test_preparation_course']
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.lower().str.strip()

print("Sample of standardized columns:")
display(df[[col for col in categorical_cols if col in df.columns]].head())
```

### 12. Create Pass/Fail Status Based on Score Average
Create a `pass_fail` column with a passing average score threshold of 40.

```python
score_cols = [col for col in ['math_score', 'reading_score', 'writing_score'] if col in df.columns]
if score_cols:
    df['average_score'] = df[score_cols].mean(axis=1)
    df['pass_fail'] = np.where(df['average_score'] >= 40, 'pass', 'fail')

display(df[['average_score', 'pass_fail']].head())
```

### 13. Filter Students with Math Scores Greater Than 80

```python
if 'math_score' in df.columns:
    high_math = df[df['math_score'] > 80]
    print(f"Found {len(high_math)} students with math scores > 80:")
    display(high_math.head())
else:
    print("math_score column is not present in this dataset.")
```

### 14. Rainfall Demonstration
This dataset does not contain rainfall data, so we demonstrate filtering rainfall with a small custom dataset.

```python
rain_data = {
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'rainfall': [8.2, 12.5, 4.0, 15.1, 7.8]
}
rain_df = pd.DataFrame(rain_data)
filtered_rain = rain_df[rain_df['rainfall'] > 10]

print("Days with rainfall > 10:")
display(filtered_rain)
```

### 15. Drop Unnecessary Columns
Let's drop the temporary `average_score` column we created because it served its purpose of generating the `pass_fail` label.

```python
if 'average_score' in df.columns:
    df = df.drop(columns=['average_score'])
    print("Successfully dropped 'average_score' column.")
```

### 16. Apply One-Hot Encoding to Categorical Columns

```python
encoding_cols = [col for col in ['gender', 'lunch'] if col in df.columns]
if encoding_cols:
    encoded_df = pd.get_dummies(df, columns=encoding_cols)
    display(encoded_df.head())
else:
    print("Target columns for encoding do not exist.")
```

### 17. Sort Students by Math Score

```python
if 'math_score' in df.columns:
    sorted_df = df.sort_values(by='math_score', ascending=False)
    display(sorted_df.head())
else:
    print("math_score column is not present to sort by.")
```
